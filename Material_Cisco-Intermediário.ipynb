# 🏗️ 1. ARQUITETURA DE REDES

Uma rede moderna é organizada em camadas para facilitar administração, manutenção e escalabilidade.

A Cisco utiliza frequentemente o modelo hierárquico de três camadas.

## Core Layer

Camada principal da rede.

Funções:

- Alta velocidade
- Encaminhamento de tráfego
- Baixa latência
- Alta disponibilidade

```text
Core
 ↓
Distribution
 ↓
Access
```

---

## Distribution Layer

Responsável pelas políticas da rede.

Funções:

- Roteamento entre VLANs
- ACLs
- QoS
- Segurança

---

## Access Layer

Camada onde os dispositivos finais são conectados.

Exemplos:

- Computadores
- Impressoras
- Telefones IP
- Access Points

---

# 🌐 2. MODELO OSI EM PROFUNDIDADE

## Camada 7 - Aplicação

Fornece serviços para aplicações.

Exemplos:

- HTTP
- HTTPS
- FTP

- DNS

---

## Camada 6 - Apresentação

Responsável pela representação dos dados.

Funções:

- Criptografia

Exemplo:

```text
JPEG
PNG
MP4
SSL/TLS
```

---

## Camada 5 - Sessão

Controla o estabelecimento e encerramento de conexões.

Exemplo:

```text
Login em um sistema
Sessão HTTPS
```

## Camada 4 - Transporte

Controla:

- Segmentação
- Controle de fluxo
- Controle de erros

Protocolos:

### TCP

Características:

- Orientado à conexão
- Confiável
- Usa ACK

Exemplo:

```text
HTTP
HTTPS
SSH
```

---

### UDP

Características:

- Sem conexão
- Mais rápido
- Sem confirmação

Exemplo:

```text
Streaming
VoIP
Jogos Online
DNS
```

---

## Camada 3 - Rede

- Endereçamento IP

Equipamentos:

```text
Router
Layer 3 Switch
```

## Camada 2 - Enlace

Responsável pelos endereços MAC.

Equipamentos:

```text
Switch
Bridge
```

---

## Camada 1 - Física

Transmissão elétrica, óptica ou rádio.

# 🔢 5. SUBNETTING

Subnetting é a divisão de uma rede em sub-redes menores.

Exemplo:

Rede original:

```text
192.168.1.0/24
```

Hosts disponíveis:

```text
254
```

Dividindo em duas redes:

```text
192.168.1.0/25
192.168.1.128/25
```

Hosts:

```text
126 por rede
```

---

## Fórmulas

### Quantidade de Hosts

```text
2^n - 2
```

Onde:

```text
n = bits de host
```

---

### Quantidade de Redes

```text
2^s
```

Onde:

```text
s = bits emprestados
```

---

# 📊 6. VLSM

Variable Length Subnet Mask.

Permite criar sub-redes com tamanhos diferentes.

Exemplo:

```text
TI → 100 hosts

RH → 30 hosts

Financeiro → 10 hosts
```

Ao invés de desperdiçar IPs, cada rede recebe apenas o necessário.

# 🏷️ 11. VLANs

VLANs criam redes lógicas independentes.

Exemplo:

```text
VLAN 10 RH

VLAN 20 Financeiro

VLAN 30 TI
```

---

## Benefícios

### Segurança

Usuários ficam isolados.

### Menor Broadcast

Menos tráfego desnecessário.

### Organização

Segmentação por departamentos.

---

## Configuração

```bash
vlan 10
name RH
```

---

# 🔗 12. TRUNKING

Permite passagem de várias VLANs em uma única conexão.

Protocolo:

```text
IEEE 802.1Q
```

Exemplo:

```text
Switch ↔ Switch
```

---

## Configuração

```bash
interface g0/1

switchport mode trunk
```

---

# 🌳 13. STP

Spanning Tree Protocol.

Evita loops em redes com switches redundantes.

Problema:

```text
Broadcast Storm
```

Consequências:

- Lentidão
- Queda da rede
- Alto uso da CPU

---

## Funcionamento

STP escolhe:

```text
Root Bridge
```

E bloqueia caminhos redundantes.

---

# 🔗 14. ETHERCHANNEL

Agrupa várias interfaces físicas em uma lógica.

Exemplo:

```text
4 links de 1Gbps

↓

4Gbps
```

Protocolos:

- PAgP
- LACP
