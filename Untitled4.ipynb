# **Manutenção e Monitoramento de Redes Locais**

---

## **Introdução**

A manutenção de redes locais é responsável por garantir que toda a infraestrutura de comunicação funcione corretamente, mantendo estabilidade, desempenho e segurança.

O trabalho do técnico de redes não envolve apenas instalar equipamentos, mas também:

- Diagnosticar falhas;
- Corrigir problemas;
- Realizar testes;
- Monitorar a infraestrutura;
- Registrar evidências técnicas;
- Prevenir novas ocorrências.

Uma rede bem estruturada é essencial para:

- Comunicação interna;
- Acesso à internet;
- Compartilhamento de arquivos;
- Funcionamento de servidores;
- Impressão em rede;
- Sistemas corporativos.

---

 ## **O Que é Manutenção de Redes**

A manutenção de redes é o conjunto de procedimentos técnicos utilizados para:

- Manter a rede funcionando;
- Identificar falhas;
- Corrigir problemas;
- Melhorar desempenho;
- Garantir estabilidade;
- Evitar interrupções.

---

# **Instalação x Manutenção**

## **Instalação**
Processo responsável por:

- Montar infraestrutura;
- Passar cabeamento;
- Configurar equipamentos;
- Definir IPs;
- Criar VLANs;
- Configurar serviços da rede.

---

## **Manutenção**
Processo responsável por:

- Verificar funcionamento;
- Corrigir falhas;
- Reparar equipamentos;
- Realizar testes;
- Monitorar estabilidade;
- Atualizar documentação.

---

# **Tipos de Manutenção**

## **Manutenção Preventiva**

Realizada antes que ocorram problemas.

### **Objetivos**
- Evitar falhas;
- Melhorar estabilidade;
- Reduzir indisponibilidade;
- Aumentar vida útil dos equipamentos.

### **Exemplos**
- Organização de cabos;
- Limpeza de equipamentos;
- Atualização de firmware;
- Testes periódicos;
- Verificação de LEDs.

---

## **Manutenção Corretiva**

Executada após uma falha acontecer.

### **Objetivos**
- Restaurar funcionamento;
- Corrigir falhas;
- Reduzir impacto na operação.

### **Exemplos**
- Troca de cabos;
- Correção de IP;
- Reinicialização de switch;
- Substituição de equipamentos.

---

## **Manutenção Diagnóstica**

Focada em descobrir a origem do problema.

### **Objetivos**
- Identificar causa raiz;
- Analisar sintomas;
- Definir solução correta.

### **Importância**
O técnico nunca deve corrigir problemas “no chute”. Toda correção precisa ser baseada em testes e evidências.

---

# **Problemas Mais Comuns em Redes**

## **Falhas Físicas**
- Cabo desconectado;
- Conector danificado;
- Porta defeituosa;
- Equipamento desligado;
- LEDs apagados.

---

## **Falhas Lógicas**
- IP incorreto;
- Máscara errada;
- Gateway inválido;
- DNS incorreto;
- Conflito de IP.

---

## **Falhas de Serviço**
- Servidor inacessível;
- Impressora fora da rede;
- Internet indisponível;
- Lentidão excessiva.

---

## **Problemas de Segurança**
- Acesso suspeito;
- Alterações indevidas;
- Tentativas de invasão;
- Bloqueios de comunicação.

---

#  **Sintoma x Causa**

## **Sintoma**
É aquilo que o usuário percebe.

### **Exemplos**
- “Sem internet”;
- “Rede lenta”;
- “Sistema não abre”;
- “Impressora não funciona”.

---

## **Causa**
É o motivo técnico do problema.

### **Exemplos**
- Cabo rompido;
- Porta defeituosa;
- Configuração incorreta;
- DNS inválido;
- Falha no roteador.

---

#  **Análise de Chamados Técnicos**

Um chamado técnico é o registro formal de um problema relacionado à infraestrutura da rede.

---

#  **Objetivos do Chamado**

- Registrar problemas;
- Organizar atendimentos;
- Priorizar falhas;
- Facilitar diagnósticos;
- Documentar ações técnicas.

---

#  **Informações Importantes em um Chamado**

## **Identificação**
- Usuário;
- Setor;
- Equipamento afetado.

---

## **Problema Relatado**
- Sintoma observado;
- Momento em que ocorreu;
- Impacto causado.

---

## **Informações Técnicas**
- Dispositivos envolvidos;
- Testes realizados;
- Possíveis causas;
- Encaminhamento técnico.

---

#  **Classificação de Prioridade**

## **Baixa**
Problema não impede o trabalho.

---

## **Média**
Afeta parcialmente um usuário ou setor.

---

## **Alta**
Afeta vários usuários ou serviços importantes.

---

## **Crítica**
Paralisa operações da empresa ou afeta segurança.

#  **Testes de Conectividade**

Os testes de conectividade servem para confirmar:

- Comunicação entre dispositivos;
- Funcionamento de serviços;
- Estabilidade da rede;
- Correção das falhas.

---

#  **Verificações Físicas Iniciais**

Antes dos comandos, o técnico deve verificar:

- Cabos conectados;
- Equipamentos ligados;
- LEDs ativos;
- Portas funcionando;
- Conectores encaixados;
- Outros dispositivos afetados.

---

#  **Ferramentas e Comandos Utilizados**

## **ping**
Verifica comunicação entre dispositivos.

---

## **ipconfig**
Mostra configurações IP do equipamento.

---

## **tracert**
Mostra o caminho da comunicação na rede.

---

## **Teste em Outro Ponto**
Confirma se o problema é local ou geral.

---

## **Teste com Outro Cabo**
Verifica possíveis danos físicos.

---

#  **Validação das Correções**

Após aplicar uma correção, o técnico precisa comprovar que o problema foi resolvido.

---

#  **Objetivos da Validação**

- Confirmar funcionamento;
- Comparar antes e depois;
- Registrar evidências;
- Garantir estabilidade.

---

#  **Exemplos de Validação**

- Ping respondendo;
- Internet funcionando;
- Impressora acessível;
- Servidor online;
- Comunicação restabelecida.

---

#  **Causa Raiz**

A causa raiz é o verdadeiro motivo do problema.

## **Exemplo**

### **Sintoma**
Computador sem internet.

### **Possíveis causas**
- Cabo desconectado;
- IP incorreto;
- Porta defeituosa;
- Gateway inválido.

---

#  **Plano de Testes**

Documento técnico utilizado para organizar verificações da infraestrutura.

---

#  **O Que Deve Conter**

- Item testado;
- Ferramenta utilizada;
- Resultado esperado;
- Resultado obtido;
- Situação final;
- Evidências registradas.

---

#  **Critérios de Aprovação**

Um teste pode ser considerado aprovado quando:

- Funcionou corretamente;
- Não apresentou falhas;
- Resultado foi registrado;
- Evidências foram documentadas.

---

#  **Situações Possíveis**

## **Aprovado**
Funcionou conforme esperado.

---

## **Reprovado**
Apresentou falha.

---

## **Pendente**
Não foi possível concluir o teste.

---

## **Em Observação**
Funcionou, mas necessita acompanhamento.

---

# ## **Homologação da Rede**

A homologação confirma que toda a infraestrutura está funcionando corretamente.

---

#  **Diferença Entre Validação e Homologação**

## **Validação**
Confirma uma correção específica.

---

## **Homologação**
Confirma funcionamento completo da rede.

---

# ## **Itens Normalmente Homologados**

- Internet;
- Gateway;
- DNS;
- Servidores;
- Impressoras;
- Comunicação interna;
- Switches;
- Roteadores.

---

#  **Monitoramento de Redes**

Monitoramento é o acompanhamento contínuo da infraestrutura da rede.

Seu objetivo é detectar falhas antes que causem impactos maiores.

---

#  **Objetivos do Monitoramento**

- Identificar lentidão;
- Detectar quedas;
- Registrar eventos;
- Melhorar estabilidade;
- Evitar interrupções;
- Analisar comportamento da rede.

---

#  **Importância do Monitoramento**

Uma rede funcionando agora não garante funcionamento futuro.

Por isso, o monitoramento deve ser contínuo.

---

#  **Teste Manual x Monitoramento**

| Teste Manual | Monitoramento |
|---|---|
| Mostra situação atual | Mostra histórico |
| Executado manualmente | Automatizado |
| Exemplo: ping | Exemplo: Zabbix |
| Momentâneo | Contínuo |

---

#  **Ferramentas de Monitoramento**

## **Syslog**
Registro de eventos da rede.

---

## **Zabbix**
Monitoramento completo de dispositivos e serviços.

---

## **PRTG**
Monitoramento visual da infraestrutura.

---

#  **Eventos, Falhas e Incidentes**

## **Evento**
Qualquer atividade registrada na rede.

### **Exemplos**
- Porta do switch caiu;
- Login realizado;
- Servidor reiniciado.

---

## **Falha**
Problema técnico que prejudica funcionamento.

### **Exemplos**
- Impressora sem resposta;
- Cabo danificado;
- DNS inválido.

---

## **Incidente**
Falha que afeta usuários ou serviços.

### **Exemplos**
- Setor inteiro sem internet;
- Sistema indisponível.

---

## **Reincidência**
Quando o mesmo problema volta a acontecer.

---

#  **Monitoramento Pós-Manutenção**

Mesmo após resolver falhas, a infraestrutura deve continuar sendo acompanhada.

---

# ## **Objetivos**

- Evitar reincidência;
- Detectar novas falhas;
- Acompanhar estabilidade;
- Atualizar documentação;
- Apoiar futuras manutenções.

---

#  **Itens Importantes Para Monitoramento**

- Gateway;
- Internet;
- Servidores;
- Switches;
- Impressoras;
- Comunicação entre setores.

---

#  **Sinais de Alerta**

- Lentidão frequente;
- Queda de conexão;
- IP duplicado;
- LEDs apagados;
- Chamados repetidos;
- Servidor inacessível.

---

---

#  **Objetivos da Documentação**
