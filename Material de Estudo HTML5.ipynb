# 🌎 INTRODUÇÃO AO HTML

HTML é a linguagem responsável pela estruturação da internet.

Praticamente todo site usa HTML:

- YouTube
- Google
- GitHub
- Netflix
- Instagram
- Discord Web
- Spotify Web

---

# 📌 O QUE É HTML?

HTML significa:

> HyperText Markup Language

Tradução:

> Linguagem de Marcação de Hipertexto

---

# ⚠️ HTML NÃO É LINGUAGEM DE PROGRAMAÇÃO

HTML:

❌ Não faz cálculos  
❌ Não possui lógica  
❌ Não cria algoritmos  
❌ Não executa decisões  

HTML apenas:

✅ Estrutura conteúdos  
✅ Organiza páginas  
✅ Define elementos

# 🏗️ ESTRUTURA BÁSICA DO HTML

```html
<!DOCTYPE html>
<html lang="pt-BR">

<head>

    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">

    <title>Meu Site</title>

</head>

<body>

    <h1>Olá Mundo</h1>

</body>

</html>
```

# 🌍 EXEMPLO DE SITE COMPLETO

```html
<!DOCTYPE html>
<html lang="pt-BR">

<head>

    <meta charset="UTF-8">

    <meta name="viewport"
    content="width=device-width, initial-scale=1.0">

    <title>Meu Site</title>

</head>

<body>

<header>

    <h1>Meu Site</h1>

</header>

<nav>

    <a href="#">Home</a>
    <a href="#">Sobre</a>

</nav>

<main>

<section>

    <h2>Sobre Mim</h2>

    <p>
        Meu nome é Nast.
    </p>

</section>

<section>

    <h2>Tecnologias</h2>

    <ul>

        <li>HTML</li>
        <li>CSS</li>
        <li>JavaScript</li>

    </ul>

</section>

</main>

<footer>

    Todos os direitos reservados.

</footer>

</body>

</html>
```
