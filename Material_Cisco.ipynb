# 🌐 1. INTRODUÇÃO ÀS REDES DE COMPUTADORES

Uma rede de computadores é um conjunto de dispositivos interligados com o objetivo de compartilhar informações, recursos e serviços.

Atualmente praticamente tudo depende de redes:

- Internet
- Aplicativos
- Jogos Online
- Streaming
- Sistemas Bancários
- Sistemas Empresariais
- Computação em Nuvem

Sem as redes de computadores não existiria comunicação digital moderna.

---

## 🎯 Objetivos de uma Rede

Uma rede é criada para:

### Compartilhamento de Arquivos

Permite que usuários acessem documentos armazenados em outros computadores ou servidores.

### Compartilhamento de Recursos

Exemplos:

- Impressoras
- Scanners
- Servidores
- Internet

### Comunicação

Permite:

- E-mails
- Chamadas VoIP
- Videoconferências
- Mensagens instantâneas

### Centralização

Empresas utilizam servidores para armazenar dados em um único local.

### Segurança

Controle de acesso e monitoramento dos usuários.

# 🏢 2. O QUE É A CISCO

A Cisco é uma das maiores fabricantes de equipamentos de redes do mundo.

Foi fundada em 1984 e se tornou referência em:

- Infraestrutura de redes
- Segurança
- Data Centers
- Cloud Computing
- Telecomunicações

A Cisco produz equipamentos utilizados por:

- Empresas
- Bancos
- Hospitais
- Universidades
- Governos
- Provedores de Internet

---

## Produtos Cisco

### Router

Responsável pelo roteamento dos pacotes entre redes.

### Switch

Responsável pela comunicação dentro da rede local.

### Firewall

Proteção contra ameaças externas.

### Access Point

Distribuição de sinal Wi-Fi.

### Servidores

Hospedagem de sistemas e serviços.

# 🧠 3. CONCEITOS FUNDAMENTAIS

## Host

Qualquer dispositivo conectado à rede.

Exemplos:

- Computadores
- Notebooks
- Celulares
- Impressoras
- Servidores

---

## Nó

Qualquer ponto capaz de enviar ou receber dados.

---

## Pacote

Unidade básica de transmissão de dados.

Quando você acessa um site, as informações são divididas em milhares de pequenos pacotes.

---

## Largura de Banda

Quantidade máxima de dados transmitidos.

Exemplos:

| Velocidade | Equivalência |
|------------|-------------|
| 100 Mbps | Rede comum |
| 1 Gbps | Rede Gigabit |
| 10 Gbps | Rede corporativa |

---

## Latência

Tempo que um pacote leva para chegar ao destino.

Medida em:

```text
Milissegundos (ms)
```

Exemplo:

```text
10 ms = excelente
50 ms = bom
100 ms = aceitável
200+ ms = ruim
```

# 🏗️ 6. MODELO OSI

O Modelo OSI possui 7 camadas.

---

## Camada 7 – Aplicação

Interface com o usuário.

Protocolos:

- HTTP
- HTTPS
- FTP
- DNS
- SMTP

---

## Camada 6 – Apresentação

Responsável pela:

- Criptografia
- Compressão
- Formatação

---

## Camada 5 – Sessão

Controla sessões entre aplicações.

---

## Camada 4 – Transporte

Responsável pela entrega dos dados.

Protocolos:

### TCP

Orientado à conexão.

Características:

- Confiável
- Controle de erros
- Confirma recebimento

---

### UDP

Não orientado à conexão.

Características:

- Mais rápido
- Menor overhead

Utilizado em:

- Jogos
- Streaming
- Chamadas

---

## Camada 3 – Rede

Responsável pelo endereçamento IP.

Equipamento principal:

```text
Router
```

---

## Camada 2 – Enlace

Responsável pelos endereços MAC.

Equipamento:

```text
Switch
```

---

## Camada 1 – Física

Transmissão física dos dados.

Exemplos:

- Cabo UTP
- Fibra Óptica
- Rádio

---

# 🔢 8. ENDEREÇAMENTO IPv4

Formato:

```text
192.168.1.10
```

Cada octeto possui:

```text
0 a 255
```

---

## Classes IPv4

### Classe A

```text
1.0.0.0 até 126.255.255.255
```

Máscara:

```text
255.0.0.0
```

---

### Classe B

```text
128.0.0.0 até 191.255.255.255
```

Máscara:

```text
255.255.0.0
```

---

### Classe C

```text
192.0.0.0 até 223.255.255.255
```

Máscara:

```text
255.255.255.0
```

---

# 📍 9. MÁSCARA DE REDE

Define qual parte é rede e qual parte é host.

Exemplo:

```text
IP:      192.168.1.50
Máscara: 255.255.255.0
```

Rede:

```text
192.168.1.0
```

Host:

```text
50
```

# 🔌 13. EQUIPAMENTOS DE REDE

## HUB

Dispositivo antigo.

Envia dados para todas as portas.

---

## SWITCH

Opera na Camada 2.

Utiliza tabela MAC.

Encaminha apenas para o destino correto.

---

## ROUTER

Opera na Camada 3.

Realiza:

- Roteamento
- NAT
- Conexão com Internet

---

## FIREWALL

Filtra tráfego.

Permite ou bloqueia conexões.

---

## ACCESS POINT

Distribui sinal Wi-Fi.

# 🔀 14. VLAN

VLAN significa:

```text
Virtual Local Area Network
```

Permite dividir uma rede física em várias redes lógicas.

---

## Exemplo

```text
VLAN 10 → RH

VLAN 20 → Financeiro

VLAN 30 → TI
```

---

## Benefícios

- Segurança
- Organização
- Redução de broadcast
- Controle de acesso

```bash
configure terminal
```

```bash
hostname SW-CORE
```

Ver interfaces:

```bash
show ip interface brief
```

Ver VLANs:

```bash
show vlan brief
```

Ver MACs:

```bash
show mac address-table
```

Ver rotas:

```bash
show ip route
```

```bash
copy running-config startup-config
```
